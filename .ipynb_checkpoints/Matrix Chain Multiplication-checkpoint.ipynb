```python
import sys
import time
import matplotlib.pyplot as plt
import seaborn as sns

def matrix_chain_order_v1(p):
    n = len(p) - 1
    m = [[0] * (n + 1) for _ in range(n + 1)]
    s = [[0] * (n + 1) for _ in range(n + 1)]
    for l in range(2, n + 1):
        for i in range(1, n - l + 2):
            j = i + l - 1
            m[i][j] = float("inf")
            for k in range(i, j):
                q = m[i][k] + m[k + 1][j] + p[i - 1] * p[k] * p[j]
                if q < m[i][j]:
                    m[i][j] = q
                    s[i][j] = k
    return m, s, n

def get_optimal_order_v1(s, i, j):
    if i == j:
        return f"A{i}"
    else:
        return f"({get_optimal_order_v1(s, i, s[i][j])} x {get_optimal_order_v1(s, s[i][j] + 1, j)})"

def display_results_v1(p):
    m, s, n = matrix_chain_order_v1(p)
    print("\nCost Table (m):")
    header = "     " + " ".join(f"{j:5d}" for j in range(1, n + 1))
    print(header)
    print("-" * len(header))
    for i in range(1, n + 1):
        row = f"{i:2d} | " + " ".join(f"{m[i][j]:5d}" if j >= i else "     " for j in range(1, n + 1))
        print(row)
    print("\nParenthesization Table (s):")
    header = "     " + " ".join(f"{j:5d}" for j in range(1, n + 1))
    print(header)
    print("-" * len(header))
    for i in range(1, n + 1):
        row = f"{i:2d} | " + " ".join(f"{s[i][j]:5d}" if j > i else "     " for j in range(1, n + 1))
        print(row)

def matrix_chain_order_v2(p):
    n = len(p) - 1
    subproblems = 0
    m = [[0] * (n + 1) for _ in range(n + 1)]
    s = [[0] * (n + 1) for _ in range(n + 1)]
    for l in range(2, n + 1):
        for i in range(1, n - l + 2):
            j = i + l - 1
            m[i][j] = float("inf")
            for k in range(i, j):
                subproblems += 1
                q = m[i][k] + m[k + 1][j] + p[i - 1] * p[k] * p[j]
                if q < m[i][j]:
                    m[i][j] = q
                    s[i][j] = k
    return m, s, n, subproblems

def display_mcm_results_v2(p):
    start_time = time.perf_counter()
    m, s, n, subproblems = matrix_chain_order_v2(p)
    end_time = time.perf_counter()
    exec_time = end_time - start_time
    space_used = sys.getsizeof(m) + sum(sys.getsizeof(r) for r in m) + sys.getsizeof(s) + sum(sys.getsizeof(r) for r in s)
    print(f"\nMinimum Cost: {m[1][n]}")
    print(f"Optimal Multiplication Order: {get_optimal_order_v1(s, 1, n)}")
    print("\n" + "=" * 55)
    print("COMPREHENSIVE COMPLEXITY STATISTICS")
    print("=" * 55)
    print(f"Measured Execution Time  : {exec_time:.8f} seconds")
    print(f"Calculated Space Usage   : {space_used} bytes")
    print(f"Total Subproblems Solved : {subproblems}")
    print("=" * 55)
    print("\nVisualizations...")
    plot_data = [row[1:] for row in m[1:]]
    plt.figure(figsize=(8, 6))
    sns.heatmap(plot_data, annot=True, fmt=".0f", cmap="Purples", xticklabels=range(1, n + 1), yticklabels=range(1, n + 1))
    plt.title("Matrix Chain Multiplication Optimal Cost Grid Matrix Map")
    plt.xlabel("End Matrix Index Target (j)")
    plt.ylabel("Start Matrix Index Origin (i)")
    plt.show()

print("Choose Input Mode:")
print("1. Default Values")
print("2. Custom Values")
choice = input("Enter your choice (1 or 2): ").strip()

if choice == "1":
    dimensions = [10, 20, 30, 40, 30]
    display_results_v1(dimensions)
    display_mcm_results_v2(dimensions)
elif choice == "2":
    dimensions = list(map(int, input("Enter matrix dimensions separated by spaces: ").split()))
    if len(dimensions) >= 2:
        display_results_v1(dimensions)
        display_mcm_results_v2(dimensions)
    else:
        print("Error: At least 2 dimensions are required.")
else:
    print("Invalid choice selected.")
```
